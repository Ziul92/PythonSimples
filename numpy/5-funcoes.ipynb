```python
import numpy as np
```

```python
array = np.array([1, 2, 3, 4, 5, 6])
array
```

```python
soma = np.add(array, 10)
soma
```

```python
subtracao = np.subtract(array, 5)
subtracao
```

```python
# Multiplicação
produto = np.multiply(array, 2)
produto
```

```python
# Divisão
divisao = np.divide(array, 2)
divisao
```

```python
# Funções trignométricas
seno = np.sin(array)
seno
```

```python
# Cosseno

cosseno = np.cos(array)
cosseno
```

```python
# Funções estatísticas
media = np.mean(array)
media
```

```python
# Desvio padrão
desvio_padrao = np.std(array)
desvio_padrao
```

```python
# Funções de comparação
comparacao_maior_que_3 = np.greater(array, 3)
comparacao_maior_que_3
```

```python
# Combinar Funções
resultado = np.sqrt(array) + np.power(array, 2)
resultado
```

```python
# Funções exponenciais
exponencial = np.exp(array)
exponencial
```

```python
# Logaritmo
logaritmo = np.log(array)
logaritmo
```

```python
# Arredondamento
arredondamento = np.round(array / 2)
arredondamento
```

```python
# Potencia
potencia = np.power(array, 3)
potencia
```

```python
soma_array = np.sum(array)
soma_array
```
